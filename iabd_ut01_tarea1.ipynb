d. (2,0 puntos) Crea una función de nombre get_tenistas que reciba como parámetro el diccionario  dic y muestre los datos del diccionario en la forma Nombre del jugador: X Grand Slams  ganados. El resultado debería ser como sigue:  
Novak Djokovic: 24 Grand Slams ganados.  
Rafael Nadal: 22 Grand Slams ganados.  
…  


```python
dic = {
    'Novak Djokovic' : 24,
    'Rafael Nadal': 22,
    'Roger Federer':20,
    'Pete Sampras':14,
    'Roy Emerson':12
}
def get_tenistas(dic):
    for nombre, titulos in dic.items():
        print(f"{nombre}: {titulos} Grand Slams ganados.")

get_tenistas(dic)
```

e. (2,0 puntos) Define la clase Pais con un "constructor" que inicialice los atributos nombre,  capital e idioma, que serán los datos a considerar en la clase. Dentro de ella, crea los métodos  get_nombre, get_capital y get_idioma, que devolverán el nombre del país, su capital y el idioma  hablado, respectivamente. A continuación, crea una instancia u objeto de la clase Pais,  asignándole los datos que desees, y haz uso de los métodos implementados para mostrar un  mensaje con la información almacenada.  

```python
class Pais:
    def __init__(self, nombre, capital, idioma):
        self.nombre = nombre
        self.capital = capital
        self.idioma = idioma

    def get_nombre(self):
        return self.nombre

    def get_capital(self):
        return self.capital

    def get_idioma(self):
        return self.idioma

mi_pais = Pais("España", "Madrid", "español")

print(f"País: {mi_pais.get_nombre()}")
print(f"Capital: {mi_pais.get_capital()}")
print(f"Idioma: {mi_pais.get_idioma()}")
```
